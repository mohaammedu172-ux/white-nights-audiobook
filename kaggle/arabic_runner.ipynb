# Mu'allaqat audio — Arabic runner
Makes the recited audio for one poem: the original verses and the modern poem-style version, in the male classical voice,
checked by two listening models, weak lines redone automatically, then joined at 80% speed with the background music.

**Settings (right panel):** Accelerator **GPU T4**, Internet **On**. Add your music dataset with **+ Add Input** (any `.mp3` in it is used).

Then **Save Version → Save & Run All**. You can close the page; results appear in the **Output** tab (`out/`):
- `imru_al_qays_original.mp3`, `imru_al_qays_modern.mp3`, `imru_al_qays_original_then_modern.mp3` (voice only)
- the same tracks with each music file mixed in (`…__<music name>.mp3`)
- `QA_REPORT.md` — which lines the judges were unsure about, to check by ear
- `problem_words_found.csv` — words the voice tends to get wrong, to add to the rulebook
